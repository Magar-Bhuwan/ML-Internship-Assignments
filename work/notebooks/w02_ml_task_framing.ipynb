## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

- My Choosen Lane is **Product Performance**
- The goal is to predict whether a product is likely to become a high-performing product based on historical data.

- This is a **Classification** task because the output is a category:

    *High Performer
    
    *Low Performer

The prediction helps identify products that deserve promotion or further optimization.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

- Revenue > 288.0

- The target variable is **High_Performance**.

- Since there is no direct column called "High Performer", I will create a proxy target.

- Products with revenue greater than 288.0 are labeled as:

    *1 = High Performer
    
    *0 = Low Performer

This target is based on a defined business rule rather than an observed label.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

- The success metric is **Accuracy**.

- Accuracy measures the percentage of products correctly classified as high-performing or low-performing.

- A model accuracy above 80% would be considered a good starting point, although the final target depends on the business requirements and class balance.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

- Each row in the dataset represents **one product**.

- The features describe that product, and the target indicates whether it is classified as a high-performing product.

```python
import pandas as pd

df = pd.read_csv("/content/Sales_Dataset_5000_Rows.csv")

df.head()
```

```python
print(df.shape)

df.info()
```

```python
df.head(10)
```

```python
threshold = df["Revenue"].median()

print(threshold)
```

```python
df["High_Performance"] = (
    df["Revenue"] > threshold
).astype(int)

df[["Revenue", "High_Performance"]].head()
```

```python
df["High_Performance"].value_counts()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule such as:

Revenue > 288.0

does not capture all factors influencing product performance.

Product success may also depend on:

- Customer engagement
- Product category
- Click-through rate
- Seasonality
- Historical trends

Machine Learning can learn complex relationships between multiple variables that are difficult to express using fixed if-else rules.

Therefore, ML provides more flexible and accurate decision support than manually written rules.
